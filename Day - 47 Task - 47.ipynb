```python
"""
Task Day-47
Topic: AI Math - Gradient Descent (Manual in Python)

Covered:
1. Vector Operations
2. Loss Calculation (MSE)
3. Manual Gradient Computation
4. Gradient Descent Optimization
5. Iterative Learning Process
6. Error Reduction Visualization
"""

import numpy as np
import matplotlib.pyplot as plt


# =====================================================
# STEP 1: Create Dataset
# =====================================================

# Relationship: y = 2x + 3

X = np.array([1, 2, 3, 4, 5], dtype=float)
y = np.array([5, 7, 9, 11, 13], dtype=float)

n = len(X)


# =====================================================
# STEP 2: Initialize Parameters
# =====================================================

weight = 0.0
bias = 0.0

learning_rate = 0.01
epochs = 1000

loss_history = []


# =====================================================
# STEP 3: Mean Squared Error Loss Function
# =====================================================

def mse_loss(actual, predicted):
    return np.mean((actual - predicted) ** 2)


# =====================================================
# STEP 4: Gradient Descent Training
# =====================================================

for epoch in range(epochs):

    # Forward Pass
    predictions = weight * X + bias

    # Loss Calculation
    loss = mse_loss(y, predictions)
    loss_history.append(loss)

    # Compute Gradients
    dw = (-2 / n) * np.sum(X * (y - predictions))
    db = (-2 / n) * np.sum(y - predictions)

    # Parameter Update
    weight = weight - learning_rate * dw
    bias = bias - learning_rate * db

    # Progress Display
    if epoch % 100 == 0:
        print(
            f"Epoch {epoch:4d} | "
            f"Loss = {loss:.6f} | "
            f"Weight = {weight:.4f} | "
            f"Bias = {bias:.4f}"
        )


# =====================================================
# STEP 5: Final Results
# =====================================================

print("\nTraining Completed")
print("-" * 40)

print(f"Final Weight : {weight:.4f}")
print(f"Final Bias   : {bias:.4f}")

final_predictions = weight * X + bias

print("\nActual Values:")
print(y)

print("\nPredicted Values:")
print(np.round(final_predictions, 2))


# =====================================================
# STEP 6: Visualization
# =====================================================

plt.figure(figsize=(8, 5))
plt.plot(loss_history)
plt.title("Loss Reduction During Gradient Descent")
plt.xlabel("Epoch")
plt.ylabel("MSE Loss")
plt.grid(True)
plt.show()


plt.figure(figsize=(8, 5))
plt.scatter(X, y, label="Actual Data")
plt.plot(X, final_predictions, label="Learned Regression Line")
plt.xlabel("X")
plt.ylabel("Y")
plt.title("Gradient Descent Linear Regression")
plt.legend()
plt.grid(True)
plt.show()
```
